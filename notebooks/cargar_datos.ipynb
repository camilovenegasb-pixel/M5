```python
import pandas as pd
import os

# 1. Rutas relativas desde notebooks/ hacia la raíz
RUTA_EXCEL = "../Base_de_datos.xlsx"
RUTA_DESTINO_CSV = "../data/raw/base_de_datos.csv"

# 2. Cargar el dataset original
print("Cargando la base de datos...")
df = pd.read_excel(RUTA_EXCEL, sheet_name="Hoja1")

# 3. Confirmar que los datos cargaron
print(f"✅ Carga exitosa: {df.shape[0]} filas y {df.shape[1]} columnas.")

# 4. Crear la carpeta data/raw si no existe y exportar el CSV
os.makedirs("../data/raw", exist_ok=True)
df.to_csv(RUTA_DESTINO_CSV, index=False)
print(f"📂 Archivo .csv exportado correctamente en: {RUTA_DESTINO_CSV}")
```
